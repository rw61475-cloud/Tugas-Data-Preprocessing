C. Menghilangkan Missing Value (nan)

1. Bagaimana kita bisa mengisi data yang kosong secara otomatis? Dengan menggunakan kelas "SimpleImputer" yang diimpor dari library sklearn.impute. Kelas ini mendefinisikan strategi pengisian nilai yang hilang pada dataset.
2. missing_values=np.nan, Digunakan untuk memberi tahu bahwa data kosong di dataset ditandai dengan NaN.
3. strategy='mean', Menentukan cara pengisian data kosong, yaitu menggunakan nilai rata-rata dari kolom tersebut. bisa juga menggunakan median, 'most_frequent' atau modus.
4. imputer.fit(x[:, 1:3]), Digunakan untuk menghitung nilai rata-rata dari kolom ke-1 sampai ke-2 (kolom indeks 1 dan 2), lalu menyimpannya di imputer.
5. imputer.transform(x[:, 1:3]), Digunakan untuk mengganti data yang kosong dengan nilai rata-rata yang sudah dihitung sebelumnya, lalu hasilnya dimasukkan kembali ke dataset.

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
imputer.fit(x[:, 1:3])
x[:, 1:3] = imputer.transform(x[:, 1:3])
```

hasil print (x) pada tahap ini

Output matriks x setelah proses SimpleImputer menunjukkan bahwa data yang kosong sudah berhasil diisi menggunakan nilai rata-rata (mean).

Pada kolom numerik:
1. Kolom usia (indeks 1) yang sebelumnya kosong sudah diisi dengan nilai rata-rata sekitar 37.73
2. Kolom pendapatan (indeks 2) yang kosong juga sudah diisi dengan nilai rata-rata sekitar 6821.73

Sedangkan nilai NaN pada kolom kota (kolom 0) masih tetap ada, karena proses imputasi hanya dilakukan pada kolom numerik saja ([:, 1:3]). Selain itu, metode mean memang hanya bisa digunakan untuk data angka, bukan data teks seperti nama kota.

Nantinya, kolom kota ini akan ditangani pada tahap encoding (One-Hot Encoding).

```python
print (x)
```

D. Mengubah Data Kategorikal Menjadi Angka (Encoding Data - Atribut)
1. Bagaimana cara mengubah kolom teks atau kategori menjadi bentuk angka biner? Dengan mengimpor kelas "ColumnTransformer" untuk menerapkan transformasi khusus pada kolom tertentu, dan kelas "OneHotEncoder" untuk mengubah data kategori menjadi matriks biner (angka 0 dan 1).
2. transformers=[('encoder', OneHotEncoder(), [0])], Bagian ini mengatur bahwa proses One-Hot Encoding hanya dilakukan pada kolom pertama (kolom indeks 0), sedangkan kolom lain tidak ikut diproses dengan cara ini.
3. remainder='passthrough', Artinya kolom lain yang tidak di-encode tetap dipertahankan apa adanya, tidak dihapus atau diubah.
4. ct.fit_transform(x), Digunakan untuk mempelajari kategori yang ada di data sekaligus langsung mengubahnya menjadi bentuk angka biner (0 dan 1).
5. np.array(...), Digunakan untuk mengubah hasil transformasi menjadi array NumPy supaya lebih rapi dan siap dipakai untuk proses machine learning selanjutnya.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer (transformers=[('encoder', OneHotEncoder(), [0])], remainder='passthrough')
x = np.array(ct.fit_transform(x))
```

Penjelasan dari print (x) pad atahap ini

1. Kolom indeks 0-5 adalah hasil dari One-Hot Encoding. Awalnya 1 kolom kategori dipecah menjadi 6 kolom biner karena ada 6 jenis kategori. Nilai 1.0 artinya data termasuk kategori tersebut, sedangkan 0.0 artinya tidak.
2. Kolom indeks 6 adalah data usia asli yang tetap dipertahankan karena remainder='passthrough'. Di kolom ini juga terlihat nilai desimal karena data kosong sudah diisi dengan nilai rata-rata (mean).
3. Kolom indeks 7 adalah kolom numerik terakhir (misalnya pendapatan/harga) yang juga tidak di-encode. Nilai kosong di kolom ini juga sudah diisi otomatis menggunakan mean, sehingga muncul angka pecahan.


```python
print (x)
```

E. Mengubah Data Kategorikal Menjadi Angka (Encoding Data - Label)
1. Bagaimana cara mengubah data teks atau kategori pada variabel label (y) menjadi bentuk numerik sederhana? Kita bisa menggunakan "LabelEncoder". Berbeda dengan OneHotEncoder yang memecah data menjadi beberapa kolom biner (0 dan 1), LabelEncoder mengubah teks menjadi angka urut (seperti 0, 1, 2, dst.) di dalam satu kolom yang sama.
2. le = LabelEncoder(), Digunakan untuk membuat objek LabelEncoder yang akan dipakai untuk mengubah data teks menjadi angka.
3. y = le.fit_transform(y), Digunakan untuk mengubah data label (misalnya “Ya” atau “Tidak”) menjadi bentuk angka (0, 1, dst.) dalam satu kolom yang sama, lalu hasilnya disimpan kembali ke y.

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y = le.fit_transform(y)
```

hasil print (y) pada tahap ini

1. Output array y setelah proses LabelEncoder menunjukkan bahwa data label sudah berubah dari teks menjadi angka.
2. LabelEncoder secara otomatis mengurutkan kategori berdasarkan abjad, lalu mengubahnya menjadi angka. Biasanya, “Tidak” akan menjadi 0 dan “Ya” menjadi 1.
3. Jumlah datanya tetap sama, yaitu 30 elemen, karena hanya bentuknya saja yang berubah, bukan jumlah datanya. Hal ini penting supaya data tetap seimbang dengan dataset awal, tetapi sudah dalam bentuk angka yang bisa diproses oleh machine learning.

```python
print (y)
```

2. x_train dan y_train, Digunakan untuk menyimpan data yang dipakai untuk melatih model machine learning.
3. x_test dan y_test, Digunakan untuk menyimpan data yang dipakai untuk menguji performa model setelah dilatih.

5. random_state = 1, Digunakan supaya hasil pembagian data tetap sama setiap kali program dijalankan.

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 1)
```

G. Feature Scaling (Standardisasi Ukuran Data)

Feature scaling diperlukan supaya semua fitur numerik berada dalam skala yang seimbang. Kalau tidak, kolom dengan angka besar (misalnya pendapatan) bisa lebih dominan dibanding kolom dengan angka kecil (misalnya usia), sehingga hasil model jadi tidak akurat.
1. from sklearn.preprocessing import StandardScaler, Digunakan untuk memanggil StandardScaler, yaitu alat untuk menstandarkan data agar memiliki rata-rata 0 dan standar deviasi 1.
2. sc = StandardScaler(), Digunakan untuk membuat objek scaler yang nantinya dipakai untuk proses scaling data.
3. x_train[:, 3:] = sc.fit_transform(x_train[:, 3:]), Digunakan untuk menghitung sekaligus mengubah skala data training pada kolom numerik (mulai dari kolom ke-3 sampai terakhir). Bagian ini hanya dilakukan pada data training agar model belajar dari data tersebut.
4. x_test[:, 3:] = sc.transform(x_test[:, 3:]), Digunakan untuk mengubah skala data test menggunakan aturan yang sudah dipelajari dari data training, tanpa menghitung ulang.

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train[:, 3:] = sc.fit_transform(x_train[:, 3:])
x_test[:, 3:] = sc.transform(x_test[:, 3:])
```

Hasil dari print (x_train) pada tahap ini

1. Hasil print(x_train) setelah feature scaling menunjukkan data latih yang sudah dinormalisasi. Nilai yang sebelumnya besar seperti pendapatan atau usia sekarang berubah jadi angka kecil dalam rentang sekitar -3 sampai +3.

2. Kolom 0 sampai 2 tetap berisi 0.0 dan 1.0 karena itu hasil One-Hot Encoding, jadi tidak ikut di-scaling. Sedangkan kolom mulai dari indeks 3 sampai akhir berubah menjadi nilai desimal karena bagian ini adalah data numerik yang diproses oleh StandardScaler.

3. Angka negatif berarti nilai data tersebut di bawah rata-rata, sedangkan angka positif berarti di atas rata-rata. Kalau nilainya mendekati 0, artinya data tersebut hampir sama dengan nilai rata-rata kolomnya.

```python
print (x_train)
```

hasil dari print (x_test) pada tahap ini

1. Hasil print(x_test) setelah feature scaling menunjukkan data uji yang sudah disesuaikan dengan skala yang sama seperti data latih.

2. Kolom 0 sampai 2 tetap berisi 0.0 dan 1.0 karena itu hasil One-Hot Encoding, jadi tidak ikut diproses oleh StandardScaler. Sedangkan kolom mulai dari indeks 3 sampai akhir berubah menjadi angka desimal karena bagian ini adalah data numerik yang sudah distandardisasi.

3. Data uji harus memakai sc.transform() (bukan fit_transform) supaya skalanya mengikuti data latih, bukan menghitung ulang sendiri. Hal ini penting agar hasil pengujian tetap konsisten dan adil.
